# Practical: Simple AI Learning Model Using k-NN

**Objective:** Implement a simple AI learning model and compare its results on a toy dataset.

In this practical, we will build a very small **k-Nearest Neighbours (k-NN)** classifier.

The model will use:

- **Study Hours**
- **Practice Tests Completed**

to predict whether a student will:

- `Pass`
- `Fail`

We will then compare the model's predictions with the actual answers.

## 1. What Is k-NN?

**k-Nearest Neighbours (k-NN)** is a simple machine-learning algorithm.

When we give it a new example, it looks at the most similar examples it already knows.

For example:

```text
New student
    ↓
Find nearby students
    ↓
Look at their results
    ↓
Choose the most common result
```

If most nearby students passed, the model predicts `Pass`.

If most nearby students failed, the model predicts `Fail`.


## 2. Our Toy Dataset

Each student has two values:

```text
[study hours, practice tests]
```

For example:

```text
[1, 1]
```

means the student studied for 1 hour and completed 1 practice test.

We will use a very small dataset so that the example is easy to understand.


## 3. Import k-NN

We will use `KNeighborsClassifier` from the `scikit-learn` library.


````python
from sklearn.neighbors import KNeighborsClassifier
````

## 4. Create the Training Data

The training data contains examples that the model will learn from.


````python
X_train = [
    [1, 1],
    [2, 2],
    [3, 2],
    [6, 5],
    [7, 6],
    [8, 7]
]

y_train = [
    "Fail",
    "Fail",
    "Fail",
    "Pass",
    "Pass",
    "Pass"
]
````

### Understanding the data

`X_train` contains the input values:

```text
[study hours, practice tests]
```

`y_train` contains the correct result for each student.

For example:

```python
[1, 1]  →  "Fail"
[8, 7]  →  "Pass"
```

These examples are used to teach the model.


## 5. Create and Train the Model

We will use:

```text
k = 3
```

This means the model looks at the **3 nearest examples** before making a prediction.


````python
model = KNeighborsClassifier(n_neighbors=3)

model.fit(X_train, y_train)
````

### Understanding the code

```python
KNeighborsClassifier(n_neighbors=3)
```

creates a k-NN classifier with `k = 3`.

Then:

```python
model.fit(X_train, y_train)
```

gives the training examples to the model.

The model is now ready to make predictions.


## 6. Create a Small Test Dataset

Now we create some new students.

We also write their actual results so that we can compare them with the model's predictions.


````python
X_test = [
    [2, 1],
    [3, 3],
    [6, 6],
    [7, 5]
]

actual = [
    "Fail",
    "Fail",
    "Pass",
    "Pass"
]
````

## 7. Make Predictions

We use `predict()` to ask the model for its answers.


````python
predicted = model.predict(X_test)

print("Predictions:", predicted)
````

## 8. Compare Actual and Predicted Results

Now we compare what really happened with what the model predicted.


````python
for i in range(len(actual)):
    print("Actual:", actual[i], "| Predicted:", predicted[i])
````

### Expected Result

The output should look like:

```text
Actual: Fail | Predicted: Fail
Actual: Fail | Predicted: Fail
Actual: Pass | Predicted: Pass
Actual: Pass | Predicted: Pass
```

This lets us directly compare the model's predictions with the correct answers.


## 9. Calculate Simple Accuracy

Accuracy tells us how many predictions were correct.

We will calculate it using very simple Python code.


````python
correct = 0

for i in range(len(actual)):
    if actual[i] == predicted[i]:
        correct = correct + 1

accuracy = correct / len(actual)

print("Accuracy:", accuracy)
````

### Understanding Accuracy

If all 4 predictions are correct:

```text
Accuracy = 4 / 4 = 1.0
```

An accuracy of `1.0` means **100% correct on this small test set**.

This is only a toy dataset, so the result should not be treated as evidence that the model would always perform perfectly on real data.


## 10. Quick Recap

In this practical:

- We created a small training dataset.
- `KNeighborsClassifier` created the k-NN model.
- `fit()` gave the examples to the model.
- `predict()` predicted `Pass` or `Fail`.
- We compared the predicted results with the actual results.
- We calculated a simple accuracy.

> **Main idea:** k-NN predicts a new example by looking at nearby examples that it already knows.
