# Practical: Value Iteration in a Simple Grid World

**Objective:** Implement value iteration in a grid-world or similar sequential decision-making environment.

In this practical, we will use a very small **2 × 2 grid world**.

The agent starts at `A` and wants to reach the goal `G`.

```text
A ---- B
|      |
|      |
C ---- G
```

The agent receives:

- `-1` for moving to a normal state
- `+10` for reaching the goal

We will use **Value Iteration** to find which states are more useful and then choose the best action from each state.

## 1. What Is Sequential Decision Making?

In a sequential decision problem, an agent makes one decision after another.

For example:

```text
Start at A
    ↓
Choose an action
    ↓
Move to another state
    ↓
Choose another action
    ↓
Reach the goal
```

A good decision now can help the agent receive a better reward later.


## 2. Our Small Grid World

We have four states:

```text
A ---- B
|      |
|      |
C ---- G
```

`G` is the goal state.

The possible moves are:

```text
A → Right → B
A → Down  → C

B → Left  → A
B → Down  → G

C → Up    → A
C → Right → G
```

Once the agent reaches `G`, the problem is finished.


## 3. Store the States, Actions, and Rewards

We will store the possible moves in a Python dictionary.


````python
transitions = {
    'A': {'Right': 'B', 'Down': 'C'},
    'B': {'Left': 'A', 'Down': 'G'},
    'C': {'Up': 'A', 'Right': 'G'},
    'G': {}
}

rewards = {
    'A': -1,
    'B': -1,
    'C': -1,
    'G': 10
}

gamma = 0.9
````

### Understanding the code

For example:

```python
'A': {'Right': 'B', 'Down': 'C'}
```

means:

- From `A`, moving Right takes us to `B`.
- From `A`, moving Down takes us to `C`.

The `rewards` dictionary tells us the reward for reaching each state.

```python
'G': 10
```

means reaching the goal gives a reward of `10`.

`gamma = 0.9` is called the **discount factor**.

It tells the agent that future rewards are important, but slightly less important than immediate rewards.


## 4. Start with Value 0

Value Iteration gives every state a **value**.

A higher value means the state is better for reaching useful future rewards.

At the beginning, we do not know the values, so we start them all at `0`.


````python
values = {
    'A': 0,
    'B': 0,
    'C': 0,
    'G': 0
}

print(values)
````

## 5. Value Iteration

For each possible action, we calculate:

```text
new value = reward of next state + gamma × value of next state
```

Then we keep the **largest** value.

We repeat this several times so that the reward from the goal can move backwards through the grid.


````python
for i in range(5):
    new_values = values.copy()

    for state in ['A', 'B', 'C']:
        possible_values = []

        for action, next_state in transitions[state].items():
            value = rewards[next_state] + gamma * values[next_state]
            possible_values.append(value)

        new_values[state] = max(possible_values)

    values = new_values

print("Final values:")
print(values)
````

## 6. Understanding the Value Iteration Code

This loop repeats the update:

```python
for i in range(5):
```

For each state, we look at every possible action:

```python
for action, next_state in transitions[state].items():
```

Then we calculate:

```python
value = rewards[next_state] + gamma * values[next_state]
```

This combines:

- the reward we get after moving,
- the future value of the next state.

Finally:

```python
new_values[state] = max(possible_values)
```

keeps the best possible value.

We do not update `G` because it is the terminal goal state.


## 7. Find the Best Policy

A **policy** tells the agent which action to choose in each state.

Now that we have the state values, we can choose the action that gives the highest value.


````python
policy = {}

for state in ['A', 'B', 'C']:
    best_action = None
    best_value = -100

    for action, next_state in transitions[state].items():
        value = rewards[next_state] + gamma * values[next_state]

        if value > best_value:
            best_value = value
            best_action = action

    policy[state] = best_action

print("Best policy:")
print(policy)
````

## 8. Understanding the Policy

For every state, the program checks all available actions.

This part:

```python
if value > best_value:
    best_value = value
    best_action = action
```

means:

> If this action gives a better value than the previous action, remember it.

The final `policy` stores the best action for each state.


## 9. Show the Recommended Actions

Let us print the policy clearly.


````python
for state, action in policy.items():
    print("From", state, "choose", action)
````

### One Possible Result

The result will guide the agent toward the goal.

For example:

```text
From A choose Right
From B choose Down
From C choose Right
```

Because both routes from `A` can reach the goal in the same number of steps, choosing `Down` from `A` can also be equally good depending on how equal choices are handled.

The important point is that the policy guides the agent toward `G`.


## 10. Quick Recap

In this practical:

- `A`, `B`, `C`, and `G` are the **states**.
- The moves are the **actions**.
- The numbers in `rewards` are the **rewards**.
- `gamma` controls how much future rewards matter.
- **Value Iteration** repeatedly updates state values.
- The **policy** chooses the action with the highest value.

> **Main idea:** Value Iteration works backwards from future rewards and helps the agent choose actions that lead toward the best long-term result.
