# Practical 5: Simple Rule-Based Expert System

**Objective:** Build a small rule-based expert system for an advisory problem.

In this practical, we will build a very small **Laptop Troubleshooting Expert System**.

The program will ask a few simple questions and use **IF-THEN rules** to give advice.

We will keep the code very small so that you can type and understand the whole program during the lab.

## 1. What Is a Rule-Based Expert System?

A rule-based expert system uses simple rules to make a decision.

A rule usually looks like this:

```text
IF something is true
THEN give some advice
```

For our laptop advisor, we will use these rules:

```text
IF the laptop does not turn on
THEN check the charger and power cable.

IF the laptop turns on but the internet does not work
THEN check Wi-Fi and restart the router.

IF the laptop turns on and the internet works but the laptop is slow
THEN close unused programs and restart the laptop.

OTHERWISE
the system does not find a common problem.
```

> **Remember:** The answers given by the user are the facts. The `if` and `elif` statements are the rules.

## 2. Build the Expert System

We will put our rules inside one Python function.

````python
def laptop_expert(turns_on, internet_works, is_slow):
    if turns_on == "no":
        return "Check power cable of the laptop"
    elif internet_works == "no":
        return "Check WiFi and restart the router"
    elif is_slow == "yes":
        return "Close unused programs and restart the system"
    else:
        return "No common problem in the expert system"
````

## 3. Understanding the Code

The function receives three facts:

```python
turns_on
internet_works
is_slow
```

Then Python checks the rules from top to bottom.

For example:

```python
if turns_on == "no":
    return "Check the charger and power cable."
```

This means:

> **IF** the laptop does not turn on, **THEN** advise the user to check the charger and power cable.

`elif` means **otherwise, check another rule**.

`else` is used when none of the earlier rules match.

The first matching rule gives the advice.

## 4. Ask the User for Facts

Now we ask the user three questions.

Type only `yes` or `no`.

````python
turns_on = input("Does the laptop turn on? yes/no: ").lower()
internet_works = input("Does the internet work? yes/no: ").lower()
is_slow = input("Is the laptop slow? yes/no: ").lower()

advice = laptop_expert(turns_on, internet_works, is_slow)
print("\nExpert System Advice")
print(advice)
````

## 5. How the Complete System Works

The complete process is very simple:

```text
User answers questions
        ↓
Python stores the answers as facts
        ↓
The IF-THEN rules are checked
        ↓
The first matching rule gives advice
```

For example, if the user enters:

```text
Does the laptop turn on? yes
Does the internet work? no
Is the laptop slow? no
```

the system gives:

```text
Check Wi-Fi and restart the router.
```
