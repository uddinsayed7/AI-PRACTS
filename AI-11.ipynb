# Practical11: Propositional Logic and Logical Inference

**Objective:** Represent a real-world problem using propositional logic and perform logical inference to derive conclusions using an AI reasoning approach.

In this practical, we will build a very small **Weather Advisory System**.

The system will start with simple facts such as:

```text
It is raining.
```

Then it will use logical rules such as:

```text
IF it is raining
THEN the road is wet.
```

From these facts and rules, the program will derive new conclusions.

## 1. What Is Propositional Logic?

Propositional logic represents statements that can be either:

- `True`
- `False`

For example:

```text
raining
road_wet
drive_carefully
```

Each statement represents one fact.

We can also create logical rules.

For example:

```text
IF raining THEN road_wet
```

This means:

> If `raining` is true, we can conclude that `road_wet` is also true.


## 2. Our Real-World Problem

We will use this small weather advisory problem.

Initial fact:

```text
raining
```

Rules:

```text
IF raining
THEN road_wet

IF road_wet
THEN drive_carefully

IF drive_carefully
THEN reduce_speed
```

So the reasoning process should be:

```text
raining
   ↓
road_wet
   ↓
drive_carefully
   ↓
reduce_speed
```


## 3. Store the Initial Fact

We will store known facts in a Python `set`.


````python
facts = {"raining"}

print("Initial facts:", facts)
````

### Understanding the code

```python
facts = {"raining"}
```

means that the system currently knows only one thing:

```text
It is raining.
```

As the reasoning process continues, new facts will be added to this set.


## 4. Store the Logical Rules

Each rule has two parts:

```text
condition → conclusion
```

For example:

```text
raining → road_wet
```


````python
rules = {
    "raining": "road_wet",
    "road_wet": "drive_carefully",
    "drive_carefully": "reduce_speed"
}
````

### Understanding the rules

The first rule:

```python
"raining": "road_wet"
```

means:

```text
IF raining is true
THEN road_wet is true
```

The other rules work in the same way.


## 5. Perform Logical Inference

We will now use **forward chaining**.

Forward chaining starts with the facts we already know and repeatedly applies rules to derive new facts.


````python
changed = True

while changed:
    changed = False

    for condition, conclusion in rules.items():
        if condition in facts and conclusion not in facts:
            facts.add(conclusion)
            changed = True
````

## 6. Understanding the Inference Code

We start with:

```python
changed = True
```

This tells Python to begin checking the rules.

Then:

```python
if condition in facts and conclusion not in facts:
```

means:

> If the condition is already known, and the conclusion is not known yet, we can add the conclusion.

This line:

```python
facts.add(conclusion)
```

adds the new conclusion to our known facts.

The loop continues until no new facts can be added.


## 7. Display the Conclusions


````python
print("Facts after inference:")

for fact in facts:
    print(fact)
````

### Expected Result

The system should finally know:

```text
raining
road_wet
drive_carefully
reduce_speed
```

The system started only with:

```text
raining
```

but used logical inference to derive the other conclusions.


## 8. How the AI Reasoning Works

The reasoning process is:

```text
Known fact:
raining

Rule 1:
raining → road_wet

New fact:
road_wet

Rule 2:
road_wet → drive_carefully

New fact:
drive_carefully

Rule 3:
drive_carefully → reduce_speed

Final new fact:
reduce_speed
```

This is called **forward chaining** because the system starts with known facts and moves forward through the rules.


## 9. Quick Recap

In this practical:

- We represented real-world statements as **propositions**.
- `facts` stored statements that are known to be true.
- `rules` represented simple logical implications.
- The program used **forward chaining**.
- New conclusions were added when their rule conditions became true.

> **Main idea:** Logical inference allows an AI system to start from known facts and use rules to derive new conclusions.
